```python
import os
import json as _json
from pathlib import Path
from dataclasses import dataclass, field
from typing import cast
from typing import List, Optional, Dict, Any, Tuple, Union, Callable, Set
import pandas as pd
import polars as pl
import io
import numpy as np
from types import SimpleNamespace
from polars.testing import assert_frame_equal as pl_assert_frame_equal
print('pandas:', pd.__version__, ' polars:', pl.__version__)
```

```python
# ── Fixtures ────────────────────────────────────────────────────────────────

# --- exp_stats_col_assign ---
FIX_EXP_STATS_COL_ASSIGN_DATA_PD = pd.DataFrame({"claims": [1,2,3]})
FIX_EXP_STATS_COL_ASSIGN_DATA_PL = pl.from_pandas(FIX_EXP_STATS_COL_ASSIGN_DATA_PD)

# --- exp_stats_rename ---
FIX_EXP_STATS_RENAME_RENAME_DICT = {"old_name": "new_name"}

print("✅ Fixtures loaded")
```

```python
# ── Before wrappers (verbatim pandas) ───────────────────────────────────────

def before_exp_stats_col_assign(data):
    data['n_claims'] = data['claims']
    return None

def before_exp_stats_rename(rename_dict, data=None):
    if data is None:
        data = pd.DataFrame({"old_name":[1],"status":["A"]})
    data = data.rename(columns=rename_dict)
    return data
```

```python
# ── Generated wrappers (verbatim LLM-generated Polars) ──────────────────────

def gen_exp_stats_col_assign(data):

    data = data.with_columns(pl.col("claims").alias("n_claims"))
    return data

def gen_exp_stats_rename(rename_dict, data=None):
    if data is None:
        data = pl.DataFrame({"old_name":[1],"status":["A"]})

    data = data.rename(rename_dict)
    return data
```

```python
# ── Comparison helper ───────────────────────────────────────────────────────
def _index_is_trivial(idx):
    # Unnamed + integer-valued covers both a fresh RangeIndex and the leftover
    # positional index after filtering/boolean-masking a RangeIndex-based frame
    # (pandas downgrades RangeIndex to a plain Int64Index on filter, but it's
    # still just leftover row positions, not real data). A set_index(...)
    # always carries the original column's name, so any genuinely meaningful
    # index is caught by the "name is not None" branch.
    return idx.name is None and pd.api.types.is_integer_dtype(idx.dtype)


def _to_pl(r):
    if isinstance(r, pl.DataFrame): return r
    if isinstance(r, pd.DataFrame): return pl.from_pandas(r.reset_index(drop=True) if _index_is_trivial(r.index) else r.reset_index())
    if isinstance(r, pd.Series): return pl.from_pandas(r.to_frame().reset_index(drop=True) if _index_is_trivial(r.index) else r.to_frame().reset_index())
    return None

def _comparison_label(label):
    text = str(label)
    if text.lstrip().startswith(("L2", "L3")):
        return text
    return f"L2 equivalence {text}"

def compare(before_result, gen_result, label, check_row_order=False):
    raw_label = str(label)
    label_parts = raw_label.strip().split()
    is_l3 = bool(label_parts and label_parts[0].upper() == "L3")
    layer = "L3" if is_l3 else "L2"
    kind = "edge" if is_l3 else "equivalence"
    if is_l3:
        label_parts = label_parts[1:]
        if label_parts and label_parts[0].lower() in ("edge", "branch"):
            label_parts = label_parts[1:]
        display_label = " ".join(label_parts)
    else:
        display_label = raw_label

    left  = _to_pl(before_result.collect() if isinstance(before_result, pl.LazyFrame) else before_result)
    right = _to_pl(gen_result.collect() if isinstance(gen_result, pl.LazyFrame) else gen_result)
    if left is None and right is None:
        print(f"⚠️  {layer} {kind} {display_label}: both sides non-DataFrame (no output to compare)")
        return
    if left is None or right is None:
        print(f"❌ {layer} {kind} {display_label}: MISMATCH — one side returned DataFrame, other did not")
        return
    left_cols, right_cols = set(left.columns), set(right.columns)
    if left_cols != right_cols:
        print(f"❌ {layer} {kind} {display_label}: MISMATCH — column sets differ (before-only={left_cols - right_cols}, gen-only={right_cols - left_cols})")
        return
    common = list(left.columns)
    try:
        pl_assert_frame_equal(left.select(common), right.select(common),
                              check_dtypes=False, check_row_order=check_row_order)
        print(f"✅ {layer} {kind} {display_label}: MATCH")
    except Exception as e:
        print(f"❌ {layer} {kind} {display_label}: MISMATCH — {e}")
```

```python
# === Tests: exp_stats_col_assign ===

# L1 smoke – generated
try:
    _r = gen_exp_stats_col_assign(FIX_EXP_STATS_COL_ASSIGN_DATA_PL.clone())
    print("✅ L1 smoke gen_exp_stats_col_assign: OK, type=", type(_r).__name__)
except Exception as _e:
    print(f"❌ L1 smoke gen_exp_stats_col_assign: {type(_e).__name__}: {_e}")

# L1 smoke – before
try:
    _pd = FIX_EXP_STATS_COL_ASSIGN_DATA_PD.copy()
    _rb = before_exp_stats_col_assign(_pd)
    print("✅ L1 smoke before_exp_stats_col_assign: OK")
except Exception as _e:
    print(f"❌ L1 smoke before_exp_stats_col_assign: {type(_e).__name__}: {_e}")

# L2 behavioral equivalence
try:
    _pd = FIX_EXP_STATS_COL_ASSIGN_DATA_PD.copy()
    before_exp_stats_col_assign(_pd)
    _rg = gen_exp_stats_col_assign(FIX_EXP_STATS_COL_ASSIGN_DATA_PL.clone())
    compare(_pd, _rg, "exp_stats_col_assign")
except Exception as _e:
    print(f"❌ L2 equivalence exp_stats_col_assign: setup error — {type(_e).__name__}: {_e}")

# L3 edge - compare the mutation/result for an empty frame with schema.
try:
    _pd = FIX_EXP_STATS_COL_ASSIGN_DATA_PD.head(0).copy()
    before_exp_stats_col_assign(_pd)
    _rg = gen_exp_stats_col_assign(FIX_EXP_STATS_COL_ASSIGN_DATA_PL.head(0))
    compare(_pd, _rg, "L3 edge exp_stats_col_assign empty input", check_row_order=True)
except Exception as _e: print(f"❌ L3 edge exp_stats_col_assign: {type(_e).__name__}: {_e}")
```
