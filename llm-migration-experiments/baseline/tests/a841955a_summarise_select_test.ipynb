```python
# ── Fixtures ────────────────────────────────────────────────────────────────

# --- summarise_select ---
FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PD = pd.DataFrame({"coassembly": [1, 2, 3], "length": [100, 200, 300], "samples": ["s1", "s2", "s3"], "total_targets": [10, 20, 30], "total_size": [1000, 2000, 3000], "extra": ["x", "y", "z"]})
FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PL = pl.from_pandas(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PD)
FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS = FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PD

# --- summarise_unmapped ---
FIX_SUMMARISE_UNMAPPED_READ_SIZE_PATH = "/tmp/binchicken_read_size.csv"
Path(FIX_SUMMARISE_UNMAPPED_READ_SIZE_PATH).write_text("s1,100\ns2,200\ns3,300\n")
FIX_SUMMARISE_UNMAPPED_SNAKEMAKE = SimpleNamespace(input=SimpleNamespace(read_size=FIX_SUMMARISE_UNMAPPED_READ_SIZE_PATH), output=SimpleNamespace(), params=SimpleNamespace())
FIX_SUMMARISE_UNMAPPED_SUMMARY_BEFORE = pd.DataFrame({"samples": ["s1,s2", "s2", "s3"], "coassembly": ["c0", "c1", "c2"]})
FIX_SUMMARISE_UNMAPPED_SUMMARY_GEN = pl.from_pandas(FIX_SUMMARISE_UNMAPPED_SUMMARY_BEFORE)

print("✅ Fixtures loaded")
```

```python
# ── Before wrappers (verbatim pandas) ───────────────────────────────────────

def before_summarise_select(elusive_clusters):
    summary = elusive_clusters[["coassembly", "samples", "length", "total_targets", "total_size"]]
    return summary

def before_summarise_unmapped(snakemake, summary):
    read_size = pd.read_csv(snakemake.input.read_size, names = ["sample", "read_size"])
    read_sizes = read_size.set_index("sample").to_dict()["read_size"]
    summary["unmapped_size"] = summary["samples"].apply(lambda x: sum([read_sizes[sample] for sample in x.split(",")]))
    return summary
```

```python
# ── Generated wrappers (verbatim LLM-generated Polars) ──────────────────────

def gen_summarise_select(elusive_clusters):
    summary = elusive_clusters.select(["coassembly", "samples", "length", "total_targets", "total_size"])
    return summary

def gen_summarise_unmapped(snakemake, summary):
    pd = pl  # LLM used `import polars as pd`
    read_size = pd.read_csv(snakemake.input.read_size, has_header=False, new_columns=["sample", "read_size"])
    read_sizes = dict(zip(read_size["sample"].to_list(), read_size["read_size"].to_list()))
    summary = summary.with_columns(
        pd.col("samples").map_elements(lambda x: sum([read_sizes[sample] for sample in x.split(",")]), return_dtype=pd.Int64).alias("unmapped_size")
    )
    return summary
```

```python
# ── Comparison helper ───────────────────────────────────────────────────────
def _index_is_trivial(idx):
    # Unnamed + integer-valued covers both a fresh RangeIndex and the leftover
    # positional index after filtering/boolean-masking a RangeIndex-based frame
    # (pandas downgrades RangeIndex to a plain Int64Index on filter, but it's
    # still just leftover row positions, not real data). A set_index(...)
    # always carries the original column's name, so any genuinely meaningful
    # index is caught by the "name is not None" branch.
    return idx.name is None and pd.api.types.is_integer_dtype(idx.dtype)


def _to_pl(r):
    if isinstance(r, pl.DataFrame): return r
    if isinstance(r, pd.DataFrame): return pl.from_pandas(r.reset_index(drop=True) if _index_is_trivial(r.index) else r.reset_index())
    if isinstance(r, pd.Series): return pl.from_pandas(r.to_frame().reset_index(drop=True) if _index_is_trivial(r.index) else r.to_frame().reset_index())
    return None

def compare(before_result, gen_result, label, check_row_order=False):
    raw_label = str(label)
    label_parts = raw_label.strip().split()
    is_l3 = bool(label_parts and label_parts[0].upper() == "L3")
    layer = "L3" if is_l3 else "L2"
    kind = "edge" if is_l3 else "equivalence"
    if is_l3:
        label_parts = label_parts[1:]
        if label_parts and label_parts[0].lower() in ("edge", "branch"):
            label_parts = label_parts[1:]
        display_label = " ".join(label_parts)
    else:
        display_label = raw_label

    left  = _to_pl(before_result.collect() if isinstance(before_result, pl.LazyFrame) else before_result)
    right = _to_pl(gen_result.collect() if isinstance(gen_result, pl.LazyFrame) else gen_result)
    if left is None and right is None:
        print(f"⚠️  {layer} {kind} {display_label}: both sides non-DataFrame (no output to compare)")
        return
    if left is None or right is None:
        print(f"❌ {layer} {kind} {display_label}: MISMATCH — one side returned DataFrame, other did not")
        return
    left_cols, right_cols = set(left.columns), set(right.columns)
    if left_cols != right_cols:
        print(f"❌ {layer} {kind} {display_label}: MISMATCH — column sets differ (before-only={left_cols - right_cols}, gen-only={right_cols - left_cols})")
        return
    common = list(left.columns)
    try:
        pl_assert_frame_equal(left.select(common), right.select(common),
                              check_dtypes=False, check_row_order=check_row_order)
        print(f"✅ {layer} {kind} {display_label}: MATCH")
    except Exception as e:
        print(f"❌ {layer} {kind} {display_label}: MISMATCH — {e}")
```

```python
# === Tests: summarise_select ===

try:
    _r = gen_summarise_select(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PL)
    print("✅ L1 smoke gen_summarise_select: OK, type=", type(_r).__name__)
except Exception as _e:
    print(f"❌ L1 smoke gen_summarise_select: {type(_e).__name__}: {_e}")

try:
    _rb = before_summarise_select(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PD)
    print("✅ L1 smoke before_summarise_select: OK")
except Exception as _e:
    print(f"❌ L1 smoke before_summarise_select: {type(_e).__name__}: {_e}")

try:
    _rb = before_summarise_select(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PD)
    _rg = gen_summarise_select(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PL)
    compare(_rb, _rg, "summarise_select")
except Exception as _e:
    print(f"❌ L2 equivalence summarise_select: setup error — {type(_e).__name__}: {_e}")

try:
    _rb = before_summarise_select(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PD.head(0))
    _rg = gen_summarise_select(FIX_SUMMARISE_SELECT_ELUSIVE_CLUSTERS_PL.head(0))
    compare(_rb, _rg, "L3 edge summarise_select empty")
except Exception as _e:
    print(f"❌ L3 edge summarise_select empty: {type(_e).__name__}: {_e}")
```
